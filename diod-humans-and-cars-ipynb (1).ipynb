```python
# ============================================================
# CELL 1 — Install Dependencies
# ============================================================
# %%
import subprocess, sys
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q",
                       "ultralytics>=8.0.0", "supervision", "lap"])
print("✅ Dependencies installed")


# ============================================================
# CELL 2 — Imports & Paths
# ============================================================
# %%
import os, shutil, random, cv2, yaml, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from pathlib import Path
from tqdm.auto import tqdm
from PIL import Image
from collections import defaultdict
from ultralytics import YOLO

# ── Dataset (read-only, never wiped) ─────────────────────────
VISDRONE_DIR = Path("/kaggle/input/datasets/banuprasadb/visdrone-dataset/VisDrone_Dataset")
TRAIN_DIR    = VISDRONE_DIR / "VisDrone2019-DET-train"
VAL_DIR      = VISDRONE_DIR / "VisDrone2019-DET-val"
TEST_DIR     = VISDRONE_DIR / "VisDrone2019-DET-test-dev"

# ── Working dirs ──────────────────────────────────────────────
WORK_DIR  = Path("/kaggle/working")
YOLO_DATA = WORK_DIR / "yolo_dataset_v2"
RESULTS   = WORK_DIR / "results"
SAVE_DIR  = WORK_DIR / "outputs"
for d in [YOLO_DATA, RESULTS, SAVE_DIR]:
    d.mkdir(parents=True, exist_ok=True)

# ── Class mapping ─────────────────────────────────────────────
# Dataset labels are YOLO format with 10 classes (0-indexed):
#   0=pedestrian, 1=people, 2=bicycle, 3=car, 4=van,
#   5=truck, 6=tricycle, 7=awning-tricycle, 8=bus, 9=motor
#
# We remap → human=0 (pedestrian+people), car=1 (car+van)
YOLO_REMAP  = {0: 0, 1: 0, 3: 1, 4: 1}   # old_cls → new_cls
CLASS_NAMES = {0: "human", 1: "car"}
COLORS      = {0: (0, 255, 120), 1: (0, 120, 255)}   # BGR
CONF_THRESH = 0.25
IOU_THRESH  = 0.45

print("Dataset splits:")
for name, d in [("Train", TRAIN_DIR), ("Val", VAL_DIR), ("Test", TEST_DIR)]:
    imgs = len(list((d / "images").glob("*.jpg")))
    lbls = len(list((d / "labels").glob("*.txt")))
    print(f"  {name:<8}: {imgs:>5} images  {lbls:>5} labels")
print(f"\nClass mapping: {YOLO_REMAP}")
print("✅ Config ready")


# ============================================================
# CELL 3 — Dataset EDA
# ============================================================
# %%
# ── Count class distribution ──────────────────────────────────
def count_labels(split_dir, max_files=500):
    lbl_files = sorted((split_dir / "labels").glob("*.txt"))[:max_files]
    counts = defaultdict(int)
    widths, heights = [], []
    for lbl in tqdm(lbl_files, desc="Scanning", leave=False):
        with open(lbl) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 5:
                    continue
                cls = int(parts[0])
                if cls not in YOLO_REMAP:
                    continue
                counts[CLASS_NAMES[YOLO_REMAP[cls]]] += 1
                widths.append(float(parts[3]))
                heights.append(float(parts[4]))
    return dict(counts), widths, heights

print("Counting train label distribution (sample 500 files) …")
tr_counts, tr_w, tr_h = count_labels(TRAIN_DIR)
for cls, cnt in tr_counts.items():
    print(f"  {cls:<10}: {cnt:,}")
print(f"  Median box W×H : {np.median(tr_w)*100:.2f}% × {np.median(tr_h)*100:.2f}% of image")
print("  ⚠  Extremely small objects — main VisDrone challenge!")

# ── Box size distribution ─────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, data, label, color in zip(
        axes,
        [np.array(tr_w)*np.array(tr_h), tr_w, tr_h],
        ["Box Area (norm.)", "Box Width (norm.)", "Box Height (norm.)"],
        ["#4C72B0", "#DD8452", "#55A868"]):
    ax.hist(data, bins=60, color=color, alpha=0.85, edgecolor="none")
    ax.axvline(np.median(data), color="red", linestyle="--",
               label=f"Median: {np.median(data):.4f}")
    ax.set_xlabel(label); ax.set_ylabel("Freq")
    ax.set_title(f"Distribution of {label}")
    ax.legend(fontsize=9); ax.grid(alpha=0.3)
plt.suptitle("VisDrone — Bounding Box Size Distribution", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig(RESULTS / "box_distribution.png", dpi=120, bbox_inches="tight")
plt.show()
print("Saved → box_distribution.png")

# ── Sample annotated images (GT boxes) ───────────────────────
def visualize_gt(split_dir, n=6):
    img_files = random.sample(sorted((split_dir / "images").glob("*.jpg")), n)
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    for ax, img_path in zip(axes.flatten(), img_files):
        img = cv2.imread(str(img_path))
        H, W = img.shape[:2]
        lbl  = split_dir / "labels" / (img_path.stem + ".txt")
        h_cnt = c_cnt = 0
        if lbl.exists():
            with open(lbl) as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) != 5: continue
                    cls = int(parts[0])
                    if cls not in YOLO_REMAP: continue
                    new_cls = YOLO_REMAP[cls]
                    cx,cy,nw,nh = map(float, parts[1:])
                    x1=int((cx-nw/2)*W); y1=int((cy-nh/2)*H)
                    x2=int((cx+nw/2)*W); y2=int((cy+nh/2)*H)
                    cv2.rectangle(img,(x1,y1),(x2,y2),COLORS[new_cls],1)
                    if new_cls==0: h_cnt+=1
                    else: c_cnt+=1
        ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        ax.set_title(f"{img_path.name}\n👤{h_cnt}  🚗{c_cnt}", fontsize=8)
        ax.axis("off")
    plt.suptitle("VisDrone — Sample Training Images (GT Boxes)", fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig(RESULTS / "sample_annotations.png", dpi=120, bbox_inches="tight")
    plt.show()
    print("Saved → sample_annotations.png")

random.seed(42)
visualize_gt(TRAIN_DIR)

# ── Class imbalance chart ─────────────────────────────────────
val_counts, *_ = count_labels(VAL_DIR, max_files=548)
fig, ax = plt.subplots(figsize=(7, 4))
splits  = ["Train (sample)", "Val (sample)"]
humans  = [tr_counts.get("human",0), val_counts.get("human",0)]
cars    = [tr_counts.get("car",0),   val_counts.get("car",0)]
x = np.arange(2)
ax.bar(x-0.2, humans, 0.4, label="Human", color="#00cc66")
ax.bar(x+0.2, cars,   0.4, label="Car",   color="#0080ff")
ax.set_xticks(x); ax.set_xticklabels(splits)
ax.set_ylabel("Count"); ax.set_title("Class Distribution per Split")
ax.legend(); ax.grid(axis="y", alpha=0.3)
for i,(h,c) in enumerate(zip(humans,cars)):
    ax.text(i-0.2, h+50, f"{h:,}", ha="center", fontsize=8)
    ax.text(i+0.2, c+50, f"{c:,}", ha="center", fontsize=8)
plt.tight_layout()
plt.savefig(RESULTS / "class_distribution.png", dpi=120, bbox_inches="tight")
plt.show()
print("Saved → class_distribution.png")


# ============================================================
# CELL 4 — Remap Labels & Build YOLO Dataset
# ============================================================
# %%
def remap_label_file(lbl_path):
    """
    Read a 10-class YOLO label file, keep only human/car classes,
    remap to our 2-class scheme: human=0, car=1
    """
    out_lines = []
    with open(lbl_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            cls = int(parts[0])
            if cls not in YOLO_REMAP:
                continue   # skip bicycle, truck, bus, etc.
            new_cls = YOLO_REMAP[cls]
            out_lines.append(f"{new_cls} {parts[1]} {parts[2]} {parts[3]} {parts[4]}")
    return out_lines

def build_split(split_dir, out_dir):
    img_out = out_dir / "images"; img_out.mkdir(parents=True, exist_ok=True)
    lbl_out = out_dir / "labels"; lbl_out.mkdir(parents=True, exist_ok=True)
    imgs = sorted(list((split_dir/"images").glob("*.jpg")) +
                  list((split_dir/"images").glob("*.png")))
    converted = skipped = 0
    for img_path in tqdm(imgs, desc=f"{out_dir.name}", leave=False):
        lbl_src = split_dir / "labels" / (img_path.stem + ".txt")
        if not lbl_src.exists():
            skipped += 1
            continue
        lines = remap_label_file(lbl_src)
        shutil.copy(img_path, img_out / img_path.name)
        with open(lbl_out / (img_path.stem + ".txt"), "w") as f:
            f.write("\n".join(lines))
        converted += 1
    print(f"  {out_dir.name}: {converted} converted  {skipped} skipped")
    return converted

if not (YOLO_DATA / "train" / "images").exists():
    print("Building YOLO dataset …")
    n_train = build_split(TRAIN_DIR, YOLO_DATA / "train")
    n_val   = build_split(VAL_DIR,   YOLO_DATA / "val")
    n_test  = build_split(TEST_DIR,  YOLO_DATA / "test")
else:
    n_train = len(list((YOLO_DATA/"train"/"images").glob("*.jpg")))
    n_val   = len(list((YOLO_DATA/"val"/"images").glob("*.jpg")))
    print(f"Dataset already exists — train: {n_train}, val: {n_val}")

# Write YAML
yaml_path = YOLO_DATA / "visdrone.yaml"
with open(yaml_path, "w") as f:
    yaml.dump({"path": str(YOLO_DATA),
               "train": "train/images", "val": "val/images", "test": "test/images",
               "nc": 2, "names": ["human", "car"]},
              f, default_flow_style=False)
print("\nYAML config:"); print(open(yaml_path).read())

# Verify: count objects in remapped val set
h_cnt = c_cnt = 0
for lbl in (YOLO_DATA/"val"/"labels").glob("*.txt"):
    for line in open(lbl):
        if line.strip():
            cls = int(line.split()[0])
            if cls == 0: h_cnt += 1
            elif cls == 1: c_cnt += 1
print(f"✅ Val set objects — humans: {h_cnt:,}  cars: {c_cnt:,}")
print("   (Both should be large numbers — if so, remapping worked!)")


# ============================================================
# CELL 5 — Train YOLOv8s  (auto-skips if weights exist)
# ============================================================
# %%
BEST_WEIGHTS = WORK_DIR / "runs" / "visdrone_v2_fixed" / "weights" / "best.pt"
SAVED_COPY   = SAVE_DIR / "best_yolov8s_visdrone.pt"

if BEST_WEIGHTS.exists() or SAVED_COPY.exists():
    src = BEST_WEIGHTS if BEST_WEIGHTS.exists() else SAVED_COPY
    print(f"✅ Weights found at: {src} — skipping training.")
    if not BEST_WEIGHTS.exists():
        BEST_WEIGHTS.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy(SAVED_COPY, BEST_WEIGHTS)
else:
    print("Starting YOLOv8s training …")
    model = YOLO("yolov8s.pt")
    model.train(
        data          = str(yaml_path),
        epochs        = 50,
        imgsz         = 640,
        batch         = 16,
        optimizer     = "AdamW",
        lr0           = 0.001,
        lrf           = 0.01,
        warmup_epochs = 3,
        mosaic        = 1.0,
        mixup         = 0.15,
        copy_paste    = 0.1,
        degrees       = 10.0,
        scale         = 0.7,
        fliplr        = 0.5,
        hsv_h         = 0.015,
        hsv_s         = 0.7,
        hsv_v         = 0.4,
        patience      = 15,
        save          = True,
        project       = str(WORK_DIR / "runs"),
        name          = "visdrone_v2_fixed",
        exist_ok      = True,
        device        = 0,
        workers       = 4,
        verbose       = True,
    )
    shutil.copy(BEST_WEIGHTS, SAVED_COPY)
    print(f"\n✅ Training done! Saved backup → {SAVED_COPY}")

# Load model
detector   = YOLO(str(BEST_WEIGHTS))
val_images = sorted((YOLO_DATA / "val" / "images").glob("*.jpg"))
print(f"Model loaded ✅  |  Val images: {len(val_images)}")


# ============================================================
# CELL 6 — Inference Helper
# ============================================================
# %%
def run_inference(img_path, conf=CONF_THRESH, iou=IOU_THRESH):
    """Detect humans & cars. Returns annotated_bgr, human_count, car_count, result."""
    result  = detector(str(img_path), conf=conf, iou=iou, verbose=False)[0]
    img_bgr = cv2.imread(str(img_path))
    human_count = car_count = 0

    for box in result.boxes:
        cls_id = int(box.cls.item())
        conf_v = float(box.conf.item())
        x1,y1,x2,y2 = map(int, box.xyxy[0].tolist())
        color = COLORS[cls_id]
        label = f"{CLASS_NAMES[cls_id]} {conf_v:.2f}"
        cv2.rectangle(img_bgr, (x1,y1), (x2,y2), color, 2)
        (tw,th),_ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.45, 1)
        cv2.rectangle(img_bgr, (x1,y1-th-4), (x1+tw,y1), color, -1)
        cv2.putText(img_bgr, label, (x1,y1-3),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.45, (0,0,0), 1)
        if cls_id == 0: human_count += 1
        else:           car_count   += 1

    overlay = img_bgr.copy()
    cv2.rectangle(overlay, (10,10), (300,80), (15,15,15), -1)
    cv2.addWeighted(overlay, 0.6, img_bgr, 0.4, 0, img_bgr)
    cv2.putText(img_bgr, f"Humans : {human_count}", (18,40),
                cv2.FONT_HERSHEY_SIMPLEX, 0.85, (0,255,120), 2)
    cv2.putText(img_bgr, f"Cars   : {car_count}", (18,72),
                cv2.FONT_HERSHEY_SIMPLEX, 0.85, (0,120,255), 2)
    return img_bgr, human_count, car_count, result

print("✅ Inference helper ready")


# ============================================================
# CELL 7 — Training Metrics Plot
# ============================================================
# %%
csv_path = WORK_DIR / "runs" / "visdrone_v2_fixed" / "results.csv"
df_train = pd.read_csv(csv_path)
df_train.columns = df_train.columns.str.strip()

metric_map = {
    "train/box_loss"      : "Train Box Loss",
    "train/cls_loss"      : "Train Class Loss",
    "val/box_loss"        : "Val Box Loss",
    "val/cls_loss"        : "Val Class Loss",
    "metrics/mAP50(B)"    : "mAP@0.50",
    "metrics/mAP50-95(B)" : "mAP@0.50:0.95",
}
available = {k: v for k, v in metric_map.items() if k in df_train.columns}
epoch_col = df_train["epoch"] if "epoch" in df_train.columns else df_train.index
clr_list  = ["#e74c3c","#e67e22","#3498db","#9b59b6","#2ecc71","#1abc9c"]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()
for ax, (col, name), color in zip(axes, available.items(), clr_list):
    ax.plot(epoch_col, df_train[col], color=color, linewidth=2.5)
    ax.fill_between(epoch_col, df_train[col], alpha=0.1, color=color)
    best_idx = df_train[col].idxmin() if "Loss" in name else df_train[col].idxmax()
    bv = df_train[col].iloc[best_idx]; be = epoch_col.iloc[best_idx]
    ax.axvline(be, color="red", linestyle="--", alpha=0.5)
    ax.scatter([be], [bv], color="red", zorder=5, s=60)
    ax.annotate(f"Best:{bv:.4f}", xy=(be,bv), xytext=(6,4),
                textcoords="offset points", fontsize=8, color="red")
    ax.set_title(name, fontsize=12, fontweight="bold")
    ax.set_xlabel("Epoch"); ax.grid(alpha=0.3)
for ax in axes[len(available):]: ax.axis("off")

plt.suptitle("Training Metrics — YOLOv8s on VisDrone2019", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig(RESULTS / "training_metrics.png", dpi=120, bbox_inches="tight")
plt.show()
print("Saved → training_metrics.png")
print("\n📈 Final epoch metrics:")
last = df_train.iloc[-1]
for col, name in available.items():
    print(f"  {name:<25}: {last[col]:.4f}")


# ============================================================
# CELL 8 — Detection Grid (12 val images)
# ============================================================
# %%
random.seed(42)
demo_imgs = random.sample(val_images, 12)

fig, axes = plt.subplots(4, 3, figsize=(20, 26))
axes = axes.flatten()
total_h = total_c = 0
for ax, img_path in zip(axes, demo_imgs):
    ann, h, c, _ = run_inference(img_path)
    total_h += h; total_c += c
    ax.imshow(cv2.cvtColor(ann, cv2.COLOR_BGR2RGB))
    ax.set_title(f"{img_path.name}\n👤 {h}  🚗 {c}", fontsize=8)
    ax.axis("off")

plt.suptitle(f"YOLOv8s Detections — VisDrone Val Set\n"
             f"Total — Humans: {total_h}  |  Cars: {total_c}",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig(RESULTS / "detection_results.png", dpi=120, bbox_inches="tight")
plt.show()
print(f"Saved → detection_results.png")
print(f"Aggregate: {total_h} humans, {total_c} cars across {len(demo_imgs)} images")


# ============================================================
# CELL 9 — Evaluation: Precision / Recall / mAP
# ============================================================
# %%
print("Running full validation on 548 val images …")
val_metrics = detector.val(
    data=str(yaml_path), split="val",
    conf=CONF_THRESH, iou=IOU_THRESH,
    device=0, verbose=True,
)
box = val_metrics.box

print("\n" + "="*55)
print("EVALUATION RESULTS (548 val images)")
print("="*55)
print(f"  {'Metric':<22} {'Overall':>10} {'Human':>10} {'Car':>10}")
print(f"  {'─'*54}")
print(f"  {'Precision':<22} {box.mp:>10.4f} {box.p[0]:>10.4f} {box.p[1]:>10.4f}")
print(f"  {'Recall':<22} {box.mr:>10.4f} {box.r[0]:>10.4f} {box.r[1]:>10.4f}")
print(f"  {'mAP@0.50':<22} {box.map50:>10.4f} {box.ap50[0]:>10.4f} {box.ap50[1]:>10.4f}")
print(f"  {'mAP@0.50:0.95':<22} {box.map:>10.4f} {box.ap[0]:>10.4f} {box.ap[1]:>10.4f}")

# Grouped bar chart
metrics_labels = ["Precision", "Recall", "mAP@0.50", "mAP@0.50:0.95"]
human_vals   = [box.p[0],  box.r[0],  box.ap50[0], box.ap[0]]
car_vals     = [box.p[1],  box.r[1],  box.ap50[1], box.ap[1]]
overall_vals = [box.mp,    box.mr,    box.map50,   box.map]

x = np.arange(len(metrics_labels)); w = 0.25
fig, ax = plt.subplots(figsize=(12, 6))
for vals, label, color, offset in [
        (human_vals,   "Human",   "#00cc66", -w),
        (car_vals,     "Car",     "#0080ff",  0),
        (overall_vals, "Overall", "#ff6600", +w)]:
    bars = ax.bar(x+offset, vals, w, label=label,
                  color=color, alpha=0.88, edgecolor="white")
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x()+bar.get_width()/2, h+0.008,
                f"{h:.3f}", ha="center", va="bottom",
                fontsize=8.5, fontweight="bold")

ax.set_xticks(x); ax.set_xticklabels(metrics_labels, fontsize=12)
ax.set_ylim(0, 1.12); ax.set_ylabel("Score", fontsize=12)
ax.set_title("Precision / Recall / mAP — YOLOv8s on VisDrone (548 val images)",
             fontsize=13, fontweight="bold")
ax.legend(fontsize=11); ax.grid(axis="y", alpha=0.3)
ax.axhline(0.5, color="gray", linestyle="--", alpha=0.35, label="0.5 reference")
plt.tight_layout()
plt.savefig(RESULTS / "evaluation_metrics.png", dpi=120, bbox_inches="tight")
plt.show()
print("Saved → evaluation_metrics.png")


# ============================================================
# CELL 10 — Human Counting Dashboard
# ============================================================
# %%
print("Building counting dashboard …")
records = []
for img_path in tqdm(val_images[:50], desc="Counting"):
    _, h, c, _ = run_inference(img_path)
    records.append({"image": img_path.name, "humans": h, "cars": c})

df_counts = pd.DataFrame(records)
grand_h   = df_counts["humans"].sum()
grand_c   = df_counts["cars"].sum()

fig = plt.figure(figsize=(20, 11))

ax1 = fig.add_subplot(2, 2, (1, 2))
x = np.arange(len(df_counts))
ax1.bar(x-0.2, df_counts["humans"], 0.4, label="Humans", color="#00cc66", alpha=0.85)
ax1.bar(x+0.2, df_counts["cars"],   0.4, label="Cars",   color="#0080ff", alpha=0.85)
ax1.set_xticks(x)
ax1.set_xticklabels([n[:12] for n in df_counts["image"]],
                    rotation=45, ha="right", fontsize=6)
ax1.set_ylabel("Count")
ax1.set_title("Detections per Image (first 50 val images)")
ax1.legend(); ax1.grid(axis="y", alpha=0.3)

ax2 = fig.add_subplot(2, 2, 3)
ax2.pie([grand_h, grand_c], labels=["Humans","Cars"],
        colors=["#00cc66","#0080ff"], autopct="%1.1f%%", startangle=90,
        textprops={"fontsize":12},
        wedgeprops={"edgecolor":"white","linewidth":2})
ax2.set_title("Class Split")

ax3 = fig.add_subplot(2, 2, 4); ax3.axis("off")
summary = (f"COUNTING SUMMARY\n{'─'*34}\n"
           f"Images processed  : {len(df_counts)}\n"
           f"Total Humans      : {grand_h}\n"
           f"Total Cars        : {grand_c}\n"
           f"Avg Humans / img  : {df_counts['humans'].mean():.1f}\n"
           f"Avg Cars   / img  : {df_counts['cars'].mean():.1f}\n"
           f"Max Humans / img  : {df_counts['humans'].max()}\n"
           f"Max Cars   / img  : {df_counts['cars'].max()}\n"
           f"Images w/ 0 cars  : {(df_counts['cars']==0).sum()}\n"
           f"Images w/ 0 humans: {(df_counts['humans']==0).sum()}\n")
ax3.text(0.05, 0.5, summary, transform=ax3.transAxes,
         fontsize=12, va="center", fontfamily="monospace",
         bbox=dict(boxstyle="round", facecolor="#e8f4f8", alpha=0.9))

plt.suptitle("Human & Car Counting Dashboard — YOLOv8s on VisDrone",
             fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(RESULTS / "counting_dashboard.png", dpi=120, bbox_inches="tight")
plt.show()
df_counts.to_csv(RESULTS / "counting_results.csv", index=False)
print(f"Saved → counting_dashboard.png")
print(f"🎯 Grand Total → Humans: {grand_h}  |  Cars: {grand_c}")


# ============================================================
# CELL 11 — FPS Benchmark
# ============================================================
# %%
N = 50
bench = [str(p) for p in val_images[:N]]
detector(bench[0], verbose=False)   # warmup
t0 = time.perf_counter()
for p in bench:
    detector(p, conf=CONF_THRESH, iou=IOU_THRESH, verbose=False)
fps = N / (time.perf_counter() - t0)
print(f"⚡ GPU Inference: {fps:.1f} FPS  ({N} images in {N/fps:.2f}s)")


# ============================================================
# CELL 12 — BONUS: ByteTrack Object Tracking
# ============================================================
# %%
print("="*55)
print("BONUS — ByteTrack Tracking")
print("="*55)

TRACK_FRAMES  = sorted(val_images)[:80]
np.random.seed(7)
PALETTE       = np.random.randint(50, 230, size=(1000,3), dtype=np.uint8)
tracker_model = YOLO(str(BEST_WEIGHTS))
track_log     = []
annotated     = []

for frame_idx, img_path in enumerate(tqdm(TRACK_FRAMES, desc="Tracking")):
    result  = tracker_model.track(
        str(img_path), conf=0.30, iou=IOU_THRESH,
        tracker="bytetrack.yaml", persist=True, verbose=False
    )[0]
    img_bgr   = cv2.imread(str(img_path))
    human_ids = set(); car_ids = set()

    if result.boxes.id is not None:
        for box, tid in zip(result.boxes, result.boxes.id):
            tid    = int(tid.item())
            cls_id = int(box.cls.item())
            conf_v = float(box.conf.item())
            x1,y1,x2,y2 = map(int, box.xyxy[0].tolist())
            color = tuple(int(c) for c in PALETTE[tid % 1000])
            label = f"ID{tid} {CLASS_NAMES[cls_id]}"
            cv2.rectangle(img_bgr,(x1,y1),(x2,y2),color,2)
            (tw,th),_ = cv2.getTextSize(label,cv2.FONT_HERSHEY_SIMPLEX,0.45,1)
            cv2.rectangle(img_bgr,(x1,y1-th-4),(x1+tw,y1),color,-1)
            cv2.putText(img_bgr,label,(x1,y1-3),
                        cv2.FONT_HERSHEY_SIMPLEX,0.45,(255,255,255),1)
            (human_ids if cls_id==0 else car_ids).add(tid)
            track_log.append({"frame":frame_idx,"id":tid,
                               "cls":CLASS_NAMES[cls_id],
                               "x1":x1,"y1":y1,"x2":x2,"y2":y2,"conf":conf_v})

    overlay = img_bgr.copy()
    cv2.rectangle(overlay,(10,10),(330,85),(15,15,15),-1)
    cv2.addWeighted(overlay,0.55,img_bgr,0.45,0,img_bgr)
    cv2.putText(img_bgr,f"Frame  : {frame_idx:03d}",(18,35),
                cv2.FONT_HERSHEY_SIMPLEX,0.65,(200,200,200),1)
    cv2.putText(img_bgr,f"Humans : {len(human_ids)}",(18,62),
                cv2.FONT_HERSHEY_SIMPLEX,0.65,(0,255,120),2)
    cv2.putText(img_bgr,f"Cars   : {len(car_ids)}",(185,62),
                cv2.FONT_HERSHEY_SIMPLEX,0.65,(0,120,255),2)
    annotated.append(img_bgr)

df_tracks = pd.DataFrame(track_log)
print(f"\n✅ Tracked {len(TRACK_FRAMES)} frames")
print(f"   Unique human IDs : {df_tracks[df_tracks.cls=='human']['id'].nunique()}")
print(f"   Unique car IDs   : {df_tracks[df_tracks.cls=='car']['id'].nunique()}")
df_tracks.to_csv(RESULTS / "tracking_log.csv", index=False)


# ============================================================
# CELL 13 — Tracking Video + Frame Grid
# ============================================================
# %%
VIDEO_OUT = RESULTS / "tracking_output.mp4"
H, W      = annotated[0].shape[:2]
writer    = cv2.VideoWriter(str(VIDEO_OUT),
                            cv2.VideoWriter_fourcc(*"mp4v"), 5, (W,H))
for frame in annotated: writer.write(frame)
writer.release()
print(f"Video saved → {VIDEO_OUT}")

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, i in zip(axes.flatten(),
                 np.linspace(0, len(annotated)-1, 6, dtype=int)):
    ax.imshow(cv2.cvtColor(annotated[i], cv2.COLOR_BGR2RGB))
    ax.set_title(f"Frame {i}", fontsize=10); ax.axis("off")
plt.suptitle("ByteTrack — Human & Car Tracking", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(RESULTS / "tracking_frames.png", dpi=120, bbox_inches="tight")
plt.show()
print("Saved → tracking_frames.png")


# ============================================================
# CELL 14 — Trajectory Plot
# ============================================================
# %%
def plot_trajectories(df_tracks, img_shape, cls_filter="human", max_ids=20):
    df_cls  = df_tracks[df_tracks.cls==cls_filter].copy()
    if df_cls.empty:
        print(f"No {cls_filter} tracks found."); return
    df_cls["cx"] = (df_cls["x1"]+df_cls["x2"])/2
    df_cls["cy"] = (df_cls["y1"]+df_cls["y2"])/2
    top_ids = df_cls["id"].value_counts().head(max_ids).index
    H, W    = img_shape[:2]
    fig, ax = plt.subplots(figsize=(13,8))
    ax.set_xlim(0,W); ax.set_ylim(H,0)
    ax.set_facecolor("#0d1117"); fig.patch.set_facecolor("#0d1117")
    ax.set_title(f"Track Trajectories — {cls_filter.title()}s",
                 color="white", fontsize=14, pad=12)
    cmap = plt.cm.hsv
    for i, tid in enumerate(top_ids):
        tdf = df_cls[df_cls.id==tid].sort_values("frame")
        ax.plot(tdf["cx"],tdf["cy"],"-o", color=cmap(i/max(max_ids,1)),
                linewidth=1.5, markersize=3, alpha=0.85, label=f"ID {tid}")
        ax.scatter(tdf["cx"].iloc[0], tdf["cy"].iloc[0],
                   color=cmap(i/max(max_ids,1)), s=40, marker="^", zorder=5)
    ax.legend(loc="upper right", fontsize=7, fancybox=True,
              facecolor="#161b22", labelcolor="white", ncol=2)
    plt.tight_layout()
    plt.savefig(RESULTS/f"trajectories_{cls_filter}.png",
                dpi=120, bbox_inches="tight", facecolor="#0d1117")
    plt.show()
    print(f"Saved → trajectories_{cls_filter}.png")

if len(df_tracks) > 0:
    plot_trajectories(df_tracks, annotated[0].shape, "human")
    plot_trajectories(df_tracks, annotated[0].shape, "car")


# ============================================================
# CELL 15 — Final Summary
# ============================================================
# %%
shutil.copy(BEST_WEIGHTS, SAVED_COPY)

print("\n" + "="*60)
print("OUTPUT FILES")
print("="*60)
for f in sorted(RESULTS.iterdir()):
    print(f"  {f.name:<50} {f.stat().st_size/1024:>8.1f} KB")

print(f"""
{'='*60}
FINAL SUMMARY
{'='*60}
  Model           : YOLOv8s fine-tuned on VisDrone2019
  Classes         : human=0 (pedestrian+people), car=1 (car+van)
  Train images    : {n_train:,}    Val images: {n_val}
  Inference speed : {fps:.1f} FPS on T4 GPU

  METRICS (all 548 val images)
  ├─ Precision      : {box.mp:.3f}  (human: {box.p[0]:.3f}  car: {box.p[1]:.3f})
  ├─ Recall         : {box.mr:.3f}  (human: {box.r[0]:.3f}  car: {box.r[1]:.3f})
  ├─ mAP@0.50       : {box.map50:.3f}  (human: {box.ap50[0]:.3f}  car: {box.ap50[1]:.3f})
  └─ mAP@0.50:0.95  : {box.map:.3f}  (human: {box.ap[0]:.3f}  car: {box.ap[1]:.3f})

  LABEL FORMAT (KEY DISCOVERY)
  • Labels are YOLO format with 10 classes (0–9)
  • Remapped: {{0,1}} → human=0  |  {{3,4}} → car=1
  • Classes 2,5,6,7,8,9 (bicycle/truck/bus etc.) discarded

  CHALLENGES
  • Tiny objects (median ~20×20 px) — core VisDrone difficulty
  • Dense occlusion in crowd scenes reduces recall
  • Class imbalance varies by scene type

  BONUS: ByteTrack tracking across 80 frames with unique IDs
""")
```
